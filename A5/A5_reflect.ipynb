## Individual - Nikolas

### Learning experience

When I started this course, my knowledge of open source soultions (and coding in general) was very limited. I have been taught Revit for drawings, Maple for maths and Matlab for data treatment, but never for any real coding. Luckily for me, I found a group member much more confident in code than I.

Over the course of the course, I learned a lot about the basic workings of Ifc and really how versatile openBIM is, compared to Autodesks' products. And they weren't even harder to use, which was interesting.

Taking this course, however, also shows me that I have just scrathed the surface of what OpenBIM has to offer. I now know a bit about extracting some data from some types of Ifc models, but I know nothing about the modeling process, nothing about what makes a model 'good quality' and so on.

As such, I should familiarize myself further in the future, as Autodesk should not be allowed a monopoly on building drawings, and I need to be knowledgeable about this much more general and open approach to building models.

### Thoughts on the process

During the process of developing our tutorial, I thought a lot about what makes for a good tutorial, what information do we need to provide in order for another student to find this tool to be usable. In that aspect, especially our managers' solution shows some promise, as I believe there is a pretty high perceived barrier to entry towards anything to do with code.

As for choosing what we wanted to develop, I really liked the open landscape and the overview of typical areas we were given. This course in general really falls into how I believe a Master's course should be taught; by giving problems to solve, not solutions. 

I also found the amount of assignments to be in a good spot. Further, the requirement for ourselves to assess the scope of an assignment was really cool, as it translates nicely into skills needed for our jobs later on.

### My future within OpenBIM

I believe my future with OpenBIM will be restricted to understanding how it works and how to get the information I want from it. We'll see in the future how it is implemented, but it will likely depend a lot on which direction my career will be heading. If, say, I get hired to create models for construction, and OpenBIM solutions are required (i.e. no Revit), I will be forced to learn other methods. (Which is probably for the best of the industry as a whole, but that is another topic.)

My belief is, that I will need to be able to teach myself how to work within OpenBIM. And I hope I am in a company with ressources that allow me to learn this. I think that OpenBIM tools will need to become part of my repetoire. If I were, however, to compare OpenBIM and Autodesk to the computing side, I would compare them to Linux and Windows/Mac. While on paper, OpenBIM is a lot better for the industry and for a lot of other things, Revit is known and seemingly more accessible (at least while it is still being taught by default)

### Wrap-up

I will divide this into the different 'segments' of the course as I experienced it:

#### A1 & A2 - Starting steps

I work from a philosophy that I want to know why and what we do what we do before we do it, so having these sessions where we identify what we want to look into before the actual code behind it was refreshing, as it gave a really nice sense of direction for what we wanted to look into.

#### A3 & A4 - Development & Documentation

This part was honestly mostly confusing. I do get that a point is probably made that documentation should be done alongside development (seeing as the assignment deadline was the exact same day for both). It was a bit difficult to find out what we were supposed to do at a given time, but I assume it was intentional, seeing as Tim is teaching the course, and seems to be an active advocate for finding things out yourself.

As for the documentation, I think it is a really important point, especially when making content as tools that are as custom as the ones we have developed are, as the need to understand exactly what is going on is much greater.

#### A5 - Reflection

Sitting here, writing my thoughts and feelings about the course, the processes and the themes surrounding it is fairly nice, as it forces me to reflect on what we have done and what my takeaways from the course is.
